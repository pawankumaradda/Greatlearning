```python
import scipy.stats as sci 

import numpy as np
import random as rand
import pandas as pd

import statsmodels.stats.proportion as prop
```

```python

temp=rand.randrange(start=5000, stop =8000)
daily_intake = np.array([temp])



```

```python
for i in range(20):
    temp=rand.randrange(start=5000, stop =8000)
    daily_intake=np.append(daily_intake,temp)
    
daily_intake.size



```

```python
# Size of the sample = 61
daily_intake.mean()
```

```python
#null hypthesis H0 = 7725

t_statistic, p_value = sci.ttest_1samp(daily_intake,7725)

print (t_statistic, p_value)
print('sample p-test = %f' % p_value)
```

```python
# using 'wilcoxon-test'

t_statistic, p_value= sci.wilcoxon(daily_intake-7725)
print (t_statistic, p_value)
```

```python
# Two sample test

energy = np.array([
# energy expenditure in mJ and stature (0=obese, 1=lean)
[9.21, 0],
[7.53, 1],
[7.48, 1],
[8.08, 1],
[8.09, 1],
[10.15, 1],
[8.40, 1],
[10.88, 1],
[6.13, 1],
[7.90, 1],
[11.51, 0],
[12.79, 0],
[7.05, 1],
[11.85, 0],
[9.97, 0],
[7.48, 1],
[8.79, 0],
[9.69, 0],
[9.68, 0],
[7.58, 1],
[9.19, 0],
[8.11, 1]])
```

```python
group1 = energy[:,1] == 0
group1 = energy[group1][:,0]
group2 = energy[:,1] == 1
group2 = energy[group2][:,0]
```

```python
# h0 = Sample's means are same
x, y = sci.ttest_ind(group1, group2)
print(x,y)
```

```python
# paired t-test 
# pre and post-surgery energy intake
intake = np.array([
[5260, 3910],
[5470, 4220],
[5640, 3885],
[6180, 5160],
[6390, 5645],
[6515, 4680],
[6805, 5265],
[7515, 5975],
[7515, 6790],
[8230, 6900],
[8770, 7335],
])
# Seperating data into 2 groups
pre = intake[:, 0]
post = intake[:, 1]
```

```python
x , y = sci.ttest_1samp (post-pre , 0)
print (x,y)
```
