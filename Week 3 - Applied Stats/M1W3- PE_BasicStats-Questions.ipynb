# 1.Import necessary packages

```python
import numpy as np
import pandas as pd 
import seaborn as sns
import scipy.stats as stats
```

# 2.Load the file

```python
data = pd.read_csv('Inc_Exp_Data.csv')
data.info()
```

# 3.Analyze the data

```python
data.head()
```

# 4.What is the Mean Expense of a Household?

```python
mean_household_expense = data['Mthly_HH_Expense'].mean()
print ("Mean household expense =%.2f" % mean_household_expense)
```

# 5.What is the Median Household Expense?

```python
median_household_expense = data['Mthly_HH_Expense'].median()
print ("Median household expense =%.2f" % median_household_expense)
```

# 6.What is the Monthly Expense for most of the Households?

```python
mode_household_expense = data['Mthly_HH_Expense'].mode()
print ("Mode household expense =%.2f" % mode_household_expense)
```

```python
data['Highest_Qualified_Member'].value_counts()
```

# 8.Calculate IQR(difference between 75% and 25% quartile)

```python
IQR = np.percentile(data['Mthly_HH_Expense'],75) - np.percentile(data['Mthly_HH_Expense'],25)
print ("IQR =%.2f" % IQR)
```
