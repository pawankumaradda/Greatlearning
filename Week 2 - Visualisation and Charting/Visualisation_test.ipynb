```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as ply

sns.set(color_codes=True)
%matplotlib inline 

#to print the graph inline rather than in seprate tab
```

```python
auto = pd.read_csv("Automobile.csv")
#auto.head(10);
auto.shape
```

```python
auto.info()
```

```python
sns.distplot(auto['highway_mpg']);
```

```python
sns.distplot(auto['city_mpg'], kde=False, rug=True);
```

```python
sns.pairplot(auto[['engine_size','horsepower']]);
```

```python
sns.pairplot(auto[['wheel_base','engine_size','city_mpg']]);
```

```python
sns.stripplot(auto['fuel_type'],auto['horsepower'], jitter=True)
```

```python
sns.boxplot(auto['horsepower'], auto['number_of_doors'], hue=auto['fuel_type']);
```

```python
sns.barplot(auto['body_style'], auto['horsepower'], hue = auto['fuel_type']);
```

```python
sns.countplot(auto['body_style'], hue = auto['fuel_type']);
```

```python
sns.pointplot(auto['horsepower'], auto['number_of_doors'], hue=auto['fuel_type']);
```

```python
sns.catplot(x="fuel_type", 
            y="horsepower",
            hue="number_of_doors",
            col="drive_wheels",
            data=auto,
            kind="strip",
           jitter=True);
            
```
