### Preprocessing 

```python
#Reshaping the set to fit the model

X_train = np.reshape(X_train,(42000, 1024)).astype('float32')
X_test = np.reshape(X_test,(18000, 1024)).astype('float32')
X_val = np.reshape(X_val,(60000, 1024)).astype('float32')
```

```python
#Converting the output variable into categorical variable
print('Value before one hot encode', y_train[0])

#One-Hot Encode the categorical column
y_train = tf.keras.utils.to_categorical(y_train,num_classes=10)
y_val= tf.keras.utils.to_categorical(y_val,num_classes=10)
y_test= tf.keras.utils.to_categorical(y_test, num_classes=10)

print('Value after one hot encode', y_train[0])
```

```python
print('Min value:', X_train.min())
print('Max value:', X_train.max(),'\n')

#Normalizing the Data
X_train = X_train/255.0
X_val= X_val/255.0
X_test = X_test/255.0

print('Min value:', X_train.min())
print('Max value:', X_train.max(),'\n')
```

### Building the Classifying Model

```python


img_model = tf.keras.models.Sequential()
img_model.add(tf.keras.layers.BatchNormalization(input_shape=(1024,),name='Input'))
img_model.add(tf.keras.layers.Dense(256,activation='relu'))
img_model.add(tf.keras.layers.Dense(256,activation='relu'))
img_model.add(tf.keras.layers.Dense(10, activation='softmax'))

img_model.summary()

img_model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
history = img_model.fit(X_train,y_train, validation_data=(X_val, y_val), batch_size=1000,epochs=100)
```

```python
hist = pd.DataFrame(history.history)
hist['epoch'] = history.epoch
hist.tail()
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=hist['epoch'], y=hist['loss'],
                    mode='lines',
                    name='loss'))
fig.add_trace(go.Scatter(x=hist['epoch'], y=hist['val_loss'],
                    mode='lines',
                    name='val_loss'))

fig.show()
```

```python
scores=[]
score = img_model.evaluate(X_train,y_train, verbose=0)
scores.append(score)
score = img_model.evaluate(X_val,y_val, verbose=0)
scores.append(score)
score = img_model.evaluate(X_test,y_test, verbose=0)
scores.append(score)
```

```python
print('Training dataset:- Loss is ', scores[0][0], 'accuracy is ', scores[0][1])
print('Validation dataset:- Loss is ', scores[1][0], 'accuracy is ', scores[1][1])
print('Testing dataset:- Loss is ', scores[2][0], 'accuracy is ', scores[2][1])
```
