# 1.You flip a fair coin 10 times. What is the probability of getting 8 or more heads? 

```python
p= .5
n= 10 

head_count = np.arange(0,11,1);
print(head_count)

binomial = stats.binom.pmf(head_count,n,p)
print(binomial)


plt.plot(head_count, binomial,'-o')

prob=0

for i in range(8,11):
    prob = prob+binomial[i]

print (f"Probability of >=8 heads is {prob}")
```
