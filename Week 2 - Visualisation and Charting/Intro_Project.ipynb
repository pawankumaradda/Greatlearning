```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="darkgrid")
```

```python
#importing u.data dataset - Rating Information
u_cols = ['user_id', 'item_id', 'rating','timestamp'];
u_data = pd.read_csv('u.data',sep='\t', names=u_cols);
u_data.head()

```

```python
#importing u.user dataset - User Information
u_cols = ['user_id', 'age', 'sex', 'occupation', 'zip_code']
u_user = pd.read_csv('u.user', sep='|', names=u_cols,encoding='latin-1')
u_user.head()
```

```python
#importing u.item dataset - Movie Information            
u_cols = ['movie_id', 'movie_title', 'release_date','video_release_date','IMDB_URL','Unknown','Action','Adventure','Children','Comedy','Crime','Documentary',
         'Drama','Fantasy','FilmNoir','Horror','Muscial','Mystery','Romance','SciFi','Thriller','War','Western'];
u_item = pd.read_csv('u.item',sep='|', names=u_cols, encoding='latin-1'); #usecols = range(7) not used

#adding index colum to the movies dataset
index = np.arange(1,len(u_item['movie_id'])+1,1)
u_item['item_id'] = index
u_item= u_item.set_index('item_id')

u_item.head()
```

## Display the top 25 movies by average rating, as a list/series/dataframe.

```python
# Getting the items that have more 100 entries
top_reviews = u_data[['item_id']].groupby('item_id')
top_reviews.filter(lambda x: x['item_id'].count()>100)

```

```python
# Average Rating per movie by groups ans selecting the top 25
rating_avg = u_data[['item_id','rating']].groupby('item_id').mean()
rating_avg = rating_avg.sort_values(by='rating',ascending=False)
order = rating_avg.index
order
```
